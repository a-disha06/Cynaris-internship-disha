# W1D5 — Data Visualisation with Matplotlib

## 1. Import Libraries and Load Dataset

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv("aiml_training_data.csv")

print("Dataset shape:", df.shape)
display(df.head())
```

## 2. Distribution of Confidence Scores

```python
plt.figure(figsize=(8, 5))

plt.hist(df["confidence"], bins=20, edgecolor="black")

plt.xlabel("Confidence")
plt.ylabel("Frequency")
plt.title("Distribution of Confidence Scores")

plt.tight_layout()
plt.show()
```

## 3. Distribution of Intents

```python
intent_counts = df["intent"].value_counts()

plt.figure(figsize=(10, 6))
plt.bar(intent_counts.index, intent_counts.values, edgecolor="black")

plt.xlabel("Intent")
plt.ylabel("Count")
plt.title("Distribution of Intents")
plt.xticks(rotation=45, ha="right")

plt.tight_layout()
plt.show()
```

## 4. Confidence Scores by Data Point

```python
plt.figure(figsize=(8, 5))

plt.scatter(
    range(len(df)),
    df["confidence"],
    alpha=0.7,
    edgecolors="black"
)

plt.xlabel("Data Point Index")
plt.ylabel("Confidence")
plt.title("Confidence Scores by Data Point")

plt.tight_layout()
plt.show()
```

## 5. Box Plot of Confidence Scores

```python
plt.figure(figsize=(8, 5))

plt.boxplot(df["confidence"], patch_artist=True)

plt.ylabel("Confidence")
plt.title("Box Plot of Confidence Scores")

plt.tight_layout()
plt.show()
```

## 6. Line Plot of Confidence Scores

```python
plt.figure(figsize=(10, 5))

plt.plot(df.index, df["confidence"], marker="o", markersize=3)

plt.xlabel("Data Point Index")
plt.ylabel("Confidence")
plt.title("Confidence Scores Across the Dataset")
plt.grid(True, alpha=0.3)

plt.tight_layout()
plt.show()
```

## 7. Average Confidence by Intent

```python
avg_confidence = df.groupby("intent")["confidence"].mean().sort_values(ascending=False)

plt.figure(figsize=(10, 6))
plt.bar(avg_confidence.index, avg_confidence.values, edgecolor="black")

plt.xlabel("Intent")
plt.ylabel("Average Confidence")
plt.title("Average Confidence by Intent")
plt.xticks(rotation=45, ha="right")

plt.tight_layout()
plt.show()
```

## 8. Intent Distribution — Pie Chart

```python
intent_counts = df["intent"].value_counts()

plt.figure(figsize=(8, 8))

plt.pie(
    intent_counts.values,
    labels=intent_counts.index,
    autopct="%1.1f%%",
    startangle=90
)

plt.title("Percentage Distribution of Intents")
plt.tight_layout()
plt.show()
```

## 9. Visualization Dashboard

```python
fig, axes = plt.subplots(1, 2, figsize=(14, 5))

# Histogram of confidence scores
axes[0].hist(df["confidence"], bins=20, edgecolor="black")
axes[0].set_title("Confidence Distribution")
axes[0].set_xlabel("Confidence")
axes[0].set_ylabel("Frequency")

# Bar chart of intent counts
intent_counts = df["intent"].value_counts()
axes[1].bar(intent_counts.index, intent_counts.values, edgecolor="black")
axes[1].set_title("Intent Distribution")
axes[1].set_xlabel("Intent")
axes[1].set_ylabel("Count")
axes[1].tick_params(axis="x", rotation=45)

plt.tight_layout()
plt.show()
```

## 10. Summary Statistics for Visualization

```python
print("Visualization Summary")
print("---------------------")

print("Dataset shape:", df.shape)
print("Number of unique intents:", df["intent"].nunique())
print("Average confidence:", round(df["confidence"].mean(), 4))
print("Minimum confidence:", round(df["confidence"].min(), 4))
print("Maximum confidence:", round(df["confidence"].max(), 4))

print("\nIntent Counts:")
display(df["intent"].value_counts().to_frame(name="Count"))

print("\nConfidence Statistics:")
display(df["confidence"].describe().to_frame(name="Confidence"))
```
