```python
import numpy as np

print("NumPy is working!")
print("NumPy version:", np.__version__)
```

```python
# Creating a 1D NumPy array

array_1d = np.array([10, 20, 30, 40, 50])

print("1D Array:")
print(array_1d)

print("Shape of 1D array:", array_1d.shape)
```

```python
# Creating a 2D NumPy array

array_2d = np.array([
    [1, 2, 3],
    [4, 5, 6]
])

print("2D Array:")
print(array_2d)

print("Shape of 2D array:", array_2d.shape)
```

```python
# Creating a 3D NumPy array

array_3d = np.array([
    [
        [1, 2],
        [3, 4]
    ],
    [
        [5, 6],
        [7, 8]
    ]
])

print("3D Array:")
print(array_3d)

print("Shape of 3D array:", array_3d.shape)
```

```python
print("Dimensions of 1D array:", array_1d.ndim)
print("Dimensions of 2D array:", array_2d.ndim)
print("Dimensions of 3D array:", array_3d.ndim)
```

```python
print("===== NumPy Array Summary =====")

print("1D Array Shape:", array_1d.shape)
print("2D Array Shape:", array_2d.shape)
print("3D Array Shape:", array_3d.shape)

print("\n1D Dimensions:", array_1d.ndim)
print("2D Dimensions:", array_2d.ndim)
print("3D Dimensions:", array_3d.ndim)
```

```python
# Broadcasting in NumPy

numbers = np.array([
    [10, 20, 30],
    [40, 50, 60]
])

print("Original Array:")
print(numbers)

result = numbers + 5

print("\nAfter adding 5 using broadcasting:")
print(result)
```

```python
# Broadcasting using a 1D array

numbers = np.array([
    [10, 20, 30],
    [40, 50, 60]
])

addition = np.array([1, 2, 3])

result = numbers + addition

print("Original Array:")
print(numbers)

print("\nArray being added:")
print(addition)

print("\nResult after broadcasting:")
print(result)
```

```python
# Vectorized operations

marks = np.array([60, 70, 80, 90, 100])

print("Original Marks:")
print(marks)

print("\nAdd 5:")
print(marks + 5)

print("\nSubtract 5:")
print(marks - 5)

print("\nMultiply by 2:")
print(marks * 2)

print("\nDivide by 2:")
print(marks / 2)

print("\nSquare:")
print(marks ** 2)
```

```python
# Matrix multiplication

A = np.array([
    [1, 2],
    [3, 4]
])

B = np.array([
    [5, 6],
    [7, 8]
])

result = A @ B

print("Matrix A:")
print(A)

print("\nMatrix B:")
print(B)

print("\nResult of A × B:")
print(result)
```

```python
import pandas as pd

data = pd.read_csv("aiml_training_data.csv")

print(data.head())
```

```python
print("Shape:", data.shape)
print("Columns:", data.columns.tolist())

print("\nData Types:")
print(data.dtypes)
```

```python
numeric_data = data.select_dtypes(include="number")

print(numeric_data.head())
```

```python
mean_confidence = data["confidence"].mean()

print("Mean Confidence:", mean_confidence)
```

```python
std_confidence = data["confidence"].std()

print("Standard Deviation:", std_confidence)
```

```python
data["confidence_standardized"] = (
    data["confidence"] - data["confidence"].mean()
) / data["confidence"].std()

print(data[["confidence", "confidence_standardized"]].head())
```

```python
correlation = numeric_data.corr()

print("Correlation Matrix:")
print(correlation)
```
